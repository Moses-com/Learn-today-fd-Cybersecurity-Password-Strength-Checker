```python
  Cybersecurity Password Strength Checker

This project checks password security strength using Python.

   Features
- Password analysis
- Password strength checker
- Security recommendations
- Strong password generator
- Security report export
```

```python
import re
import random
import string
```

```python
def check_password_strength(password):

    score = 0
    advice = []

    # Check password length
    if len(password) >= 8:
        score += 1
    else:
        advice.append("Use at least 8 characters")

    # Check uppercase letters
    if re.search(r"[A-Z]", password):
        score += 1
    else:
        advice.append("Add uppercase letters")

    # Check lowercase letters
    if re.search(r"[a-z]", password):
        score += 1
    else:
        advice.append("Add lowercase letters")

    # Check numbers
    if re.search(r"[0-9]", password):
        score += 1
    else:
        advice.append("Add numbers")

    # Check special characters
    if re.search(r"[!@#$%^&*(),.?\":{}|<>]", password):
        score += 1
    else:
        advice.append("Add special symbols")

    # Determine strength level
    if score <= 2:
        strength = "Weak"

    elif score == 3 or score == 4:
        strength = "Medium"

    else:
        strength = "Strong"

    return strength, advice
```

```python
password = input("Enter your password: ")

strength, advice = check_password_strength(password)

print("\n====================")
print("PASSWORD ANALYSIS")
print("====================")

print(f"\nPassword Strength: {strength}")

if advice:
    print("\nSecurity Advice:")

    for item in advice:
        print("-", item)

else:
    print("\nExcellent password!")
```

```python
def generate_strong_password(length=12):

    characters = (
        string.ascii_letters +
        string.digits +
        string.punctuation
    )

    strong_password = ''.join(
        random.choice(characters)
        for i in range(length)
    )

    return strong_password


print("====================")
print("GENERATED PASSWORD")
print("====================")

print(generate_strong_password())
```

```python
with open("password_security_report.txt", "w") as file:

    file.write("PASSWORD SECURITY REPORT\n")
    file.write("========================\n\n")

    file.write(f"Password Strength: {strength}\n\n")

    if advice:

        file.write("Security Advice:\n")

        for item in advice:
            file.write(f"- {item}\n")

    else:
        file.write("Excellent password security!\n")

print("Report saved successfully!")
```
